```python
Color Detection
```

```python
Import Libraries
```

```python
import cv2
import numpy as np
from sklearn.neighbors import KNeighborsClassifier
import os
```

```python
Color Dataset
```

```python
colors = {
    "Red": [255, 0, 0],
    "Green": [0, 255, 0],
    "Blue": [0, 0, 255],
    "Yellow": [255, 255, 0],
    "Cyan": [0, 255, 255],
    "Magenta": [255, 0, 255],
    "Black": [0, 0, 0],
    "White": [255, 255, 255],
    "Gray": [128, 128, 128],
    "Orange": [255, 165, 0],
    "Pink": [255, 192, 203],
    "Purple": [128, 0, 128],
    "Brown": [139, 69, 19],
    "Sky Blue": [135, 206, 235]
}
```

```python
Training the model
```

```python
X = np.array(list(colors.values()))
y = np.array(list(colors.keys()))
model = KNeighborsClassifier(n_neighbors=1)
model.fit(X, y)
def detect_color(rgb):
    rgb = np.array(rgb).reshape(1, -1)
    return model.predict(rgb)[0]
```

```python
Inserting Image and detecting
```

```python
image_path = "sample1.jpg" 
if not os.path.exists(image_path):
    print(f"Image not found: {image_path}")
else:
    img = cv2.imread(image_path)
    if img is None:
        print("Failed to load image. Check format or path.")
    else:
        img = cv2.resize(img, (400, 400))
        def click_event(event, x, y, flags, param):
            if event == cv2.EVENT_LBUTTONDOWN:
                b, g, r = img[y, x]
                color_name = detect_color((r, g, b))
                print(f"🖼️ Pixel ({x},{y}) → RGB({r},{g},{b}) → {color_name}")
                font = cv2.FONT_HERSHEY_SIMPLEX
                cv2.putText(img, color_name, (x, y), font, 0.6, (255, 255, 255), 2)
                cv2.imshow("Image", img)

        cv2.imshow("Image", img)
        cv2.setMouseCallback("Image", click_event)
        cv2.waitKey(0)
        cv2.destroyAllWindows()

```
